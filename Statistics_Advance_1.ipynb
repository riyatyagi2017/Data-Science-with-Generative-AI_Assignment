```python
import scipy.stats as stats

# Given data
sample_mean = 105
population_mean = 100
standard_deviation = 15
sample_size = 25

# Calculate the z-score
z_score = (sample_mean - population_mean) / (standard_deviation / (sample_size ** 0.5))

# Calculate the p-value (two-tailed test)
p_value = stats.norm.sf(abs(z_score)) * 2

# Significance level
alpha = 0.05

# Compare p-value to alpha
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
n = 10  # Number of trials
p = 0.6  # Probability of success
num_samples = 1000

# Generate samples
samples = np.random.binomial(n, p, size=num_samples)

# Calculate mean and variance
sample_mean = np.mean(samples)
sample_variance = np.var(samples)

# Plot the distribution
plt.hist(samples, bins=n+1, density=True, alpha=0.5)
plt.xlabel('Number of successes')
plt.ylabel('Probability')
plt.title('Binomial Distribution (n=10, p=0.6)')
plt.show()

# Expected mean and variance
expected_mean = n * p
expected_variance = n * p * (1 - p)

print("Sample Mean:", sample_mean)
print("Sample Variance:", sample_variance)
print("Expected Mean:", expected_mean)
print("Expected Variance:", expected_variance)
```
