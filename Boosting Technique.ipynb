Q 26.Train a Gradient Boosting Classifier and visualize the ROC curve

```python
# Import required libraries
import matplotlib.pyplot as plt
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import label_binarize
from sklearn.metrics import roc_curve, auc
from sklearn.multiclass import OneVsRestClassifier

# Load dataset (using Iris dataset as an example)
data = load_iris()
X = data.data
y = data.target

# Binarize the output for ROC curve (One-vs-Rest strategy for multi-class classification)
y_bin = label_binarize(y, classes=[0, 1, 2])

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y_bin, test_size=0.2, random_state=42)

# Initialize the Gradient Boosting model
model = GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, max_depth=3)

# Apply One-vs-Rest classification
ovr_model = OneVsRestClassifier(model)

# Train the model
ovr_model.fit(X_train, y_train)

# Predict probabilities for the test set
y_prob = ovr_model.predict_proba(X_test)

# Plot ROC curve for each class
plt.figure(figsize=(10, 8))

for i in range(y_bin.shape[1]):  # Loop through each class
    fpr, tpr, _ = roc_curve(y_test[:, i], y_prob[:, i])  # Compute ROC curve
    roc_auc = auc(fpr, tpr)  # Compute AUC
    plt.plot(fpr, tpr, lw=2, label=f'Class {i} (AUC = {roc_auc:.2f})')

# Plot the random classifier line (diagonal line)
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')

# Labels and title
plt.title('Receiver Operating Characteristic (ROC) Curve for Gradient Boosting Classifier')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc='lower right')
plt.grid(True)

# Show the plot
plt.show()
```

Q29. Train an AdaBoost Classifier and analyze the effect of different learning rates

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 1. Generate a Synthetic Dataset
X, y = make_moons(n_samples=1000, noise=0.3, random_state=42)

# 2. Split Data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Define a range of learning rates
learning_rates = [0.01, 0.05, 0.1, 0.2, 0.5, 0.8, 1.0, 1.2, 1.5, 2.0]
train_accuracies = []
test_accuracies = []

# 3. Train AdaBoost with Different Learning Rates
print("Training AdaBoost with different learning rates...")
for lr in learning_rates:
    # Use a Decision Tree as the base estimator
    # A shallow tree (max_depth=1 or 2) is common for AdaBoost
    base_estimator = DecisionTreeClassifier(max_depth=1, random_state=42)

    ada_clf = AdaBoostClassifier(
        estimator=base_estimator,
        n_estimators=100,  # Number of boosting stages
        learning_rate=lr,
        random_state=42
    )
    ada_clf.fit(X_train, y_train)

    y_train_pred = ada_clf.predict(X_train)
    train_acc = accuracy_score(y_train, y_train_pred)
    train_accuracies.append(train_acc)

    y_test_pred = ada_clf.predict(X_test)
    test_acc = accuracy_score(y_test, y_test_pred)
    test_accuracies.append(test_acc)
    print(f"  Learning Rate: {lr:.2f}, Train Accuracy: {train_acc:.4f}, Test Accuracy: {test_acc:.4f}")

# 4. Analyze Results
plt.figure(figsize=(10, 6))
plt.plot(learning_rates, train_accuracies, marker='o', label='Train Accuracy')
plt.plot(learning_rates, test_accuracies, marker='x', label='Test Accuracy')
plt.title('Effect of Learning Rate on AdaBoost Classifier Accuracy')
plt.xlabel('Learning Rate')
plt.ylabel('Accuracy')
plt.grid(True)
plt.legend()
plt.xticks(learning_rates)
plt.show()

# 5. Discussion of Observations
print("\n--- Analysis of Learning Rate Effect ---")
print("Observations:")
print("- **Low Learning Rates (e.g., 0.01 - 0.1):**")
print("  - The model learns slowly. With a fixed number of estimators, it might not fully converge, potentially leading to lower accuracy compared to higher learning rates initially.")
print("  - If given enough estimators, lower learning rates can lead to more robust models and potentially better generalization by carefully adjusting the weights.")
print(f"  - In this run, for LR = 0.01, Test Accuracy: {test_accuracies[learning_rates.index(0.01)]:.4f}")

print("\n- **Medium Learning Rates (e.g., 0.2 - 1.0):**")
print("  - The model learns at a more balanced pace. This range often yields good performance, finding a good trade-off between convergence speed and generalization.")
print("  - We typically see an increase in accuracy as the learning rate increases within this range, up to a certain point.")
print(f"  - In this run, for LR = 0.50, Test Accuracy: {test_accuracies[learning_rates.index(0.5)]:.4f}")
print(f"  - In this run, for LR = 1.00, Test Accuracy: {test_accuracies[learning_rates.index(1.0)]:.4f}")


print("\n- **High Learning Rates (e.g., > 1.0):**")
print("  - The model takes larger steps, which can lead to faster convergence initially but also increase the risk of overshooting the optimal solution.")
print("  - Beyond a certain point, very high learning rates can cause the model to oscillate or even diverge, resulting in lower accuracy on both training and test sets.")
print("  - This is because the weights of misclassified samples are boosted too aggressively, potentially overemphasizing noise or outliers.")
print(f"  - In this run, for LR = 1.50, Test Accuracy: {test_accuracies[learning_rates.index(1.5)]:.4f}")
print(f"  - In this run, for LR = 2.00, Test Accuracy: {test_accuracies[learning_rates.index(2.0)]:.4f}")

print("\n- **Optimal Learning Rate:**")
print("  - The optimal learning rate is usually found through experimentation (hyperparameter tuning). It's the one that provides the best balance between training performance and generalization to unseen data.")
print("  - From the plot, we can observe that the test accuracy generally peaks around a certain learning rate before potentially dropping off.")
best_lr_idx = np.argmax(test_accuracies)
best_lr = learning_rates[best_lr_idx]
best_test_acc = test_accuracies[best_lr_idx]
print(f"  - In this specific example, the highest test accuracy of {best_test_acc:.4f} was achieved with a learning rate of {best_lr}.")
```

Q 30. Train an XGBoost Classifier for multi-class classification and evaluate using log-loss.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.metrics import log_loss, accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder

# 1. Generate a Synthetic Multi-Class Dataset
# Let's create a dataset with 4 classes
n_samples = 1000
n_features = 20
n_classes = 4
X, y = make_classification(n_samples=n_samples, n_features=n_features, n_informative=10,
                           n_redundant=5, n_classes=n_classes, random_state=42)

print(f"Dataset shape: X={X.shape}, y={y.shape}")
print(f"Number of classes: {n_classes}")
print(f"Unique classes in y: {np.unique(y)}")

# XGBoost expects integer labels starting from 0.
# sklearn's make_classification usually provides this, but it's good practice
# to ensure it if you're loading data from other sources.
# Here, we'll confirm the labels are correctly encoded.
le = LabelEncoder()
y_encoded = le.fit_transform(y)
print(f"Encoded unique classes in y: {np.unique(y_encoded)}")
print(f"Original classes mapping: {list(le.classes_)}")


# 2. Split Data
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.3, random_state=42, stratify=y_encoded)

print(f"\nTrain set shape: X_train={X_train.shape}, y_train={y_train.shape}")
print(f"Test set shape: X_test={X_test.shape}, y_test={y_test.shape}")

# 3. Train XGBoost Classifier
print("\nTraining XGBoost Classifier...")
# For multi-class classification, use 'multi:softprob' to get probabilities
# and set 'num_class' to the number of unique classes.
# Use_label_encoder=False is recommended to avoid a deprecation warning.
xgb_clf = XGBClassifier(
    objective='multi:softprob',  # Outputs probabilities for each class
    num_class=n_classes,         # Number of unique classes
    n_estimators=100,            # Number of boosting rounds
    learning_rate=0.1,           # Step size shrinkage
    use_label_encoder=False,     # Suppress deprecation warning
    eval_metric='mlogloss',      # Evaluation metric for multi-class logloss
    random_state=42
)

xgb_clf.fit(X_train, y_train)

print("XGBoost Classifier trained successfully.")

# 4. Make Predictions (Probabilities for Log-Loss)
print("\nMaking predictions...")
y_train_pred_proba = xgb_clf.predict_proba(X_train)
y_test_pred_proba = xgb_clf.predict_proba(X_test)

# Also get class labels for accuracy and classification report
y_train_pred_label = xgb_clf.predict(X_train)
y_test_pred_label = xgb_clf.predict(X_test)

# 5. Evaluate using Log-Loss
train_logloss = log_loss(y_train, y_train_pred_proba)
test_logloss = log_loss(y_test, y_test_pred_proba)

print(f"\n--- Evaluation Results ---")
print(f"Train Log-Loss: {train_logloss:.4f}")
print(f"Test Log-Loss: {test_logloss:.4f}")

# 6. Additional Evaluation
train_accuracy = accuracy_score(y_train, y_train_pred_label)
test_accuracy = accuracy_score(y_test, y_test_pred_label)

print(f"\nTrain Accuracy: {train_accuracy:.4f}")
print(f"Test Accuracy: {test_accuracy:.4f}")

print("\n--- Classification Report (Test Set) ---")
# Use the original class names if available for the report
target_names = [f'Class {i}' for i in range(n_classes)]
if hasattr(le, 'classes_'): # If LabelEncoder was used
    target_names = [str(cls) for cls in le.classes_]

print(classification_report(y_test, y_test_pred_label, target_names=target_names))

# Interpretation of Log-Loss:
print("\n--- Interpreting Log-Loss ---")
print("Log-Loss (or Cross-Entropy Loss) is a measure of the prediction uncertainty.")
print("It penalizes false classifications and, more importantly, it penalizes confident false classifications heavily.")
print("A perfect model would have a Log-Loss of 0.")
print("The closer the Log-Loss is to 0, the better the model's predictions (in terms of probabilities) are.")
print("It's particularly useful for evaluating models that output probabilities, as it considers the predicted probability distribution over classes.")
print("A higher Log-Loss indicates that the model is less confident or making more incorrect predictions with high confidence.")

# Plotting the predicted probabilities (optional)
# For simplicity, let's visualize predictions for a few test samples
num_samples_to_plot = 5
print(f"\nPredicted probabilities for first {num_samples_to_plot} test samples:")
for i in range(num_samples_to_plot):
    true_class = y_test[i]
    predicted_probs = y_test_pred_proba[i]
    predicted_class = y_test_pred_label[i]

    print(f"  Sample {i+1}:")
    print(f"    True Class: {true_class} (Original: {le.inverse_transform([true_class])[0]})")
    print(f"    Predicted Class: {predicted_class} (Original: {le.inverse_transform([predicted_class])[0]})")
    print(f"    Predicted Probabilities: {np.round(predicted_probs, 3)}")
    print(f"    Log-Loss contribution: {-np.log(predicted_probs[true_class]):.4f}")

    # Small visualization for probabilities
    plt.figure(figsize=(6, 2))
    plt.bar(range(n_classes), predicted_probs, color=['blue' if j != true_class else 'red' for j in range(n_classes)])
    plt.title(f"Sample {i+1} Probabilities (True Class: {true_class})")
    plt.xlabel("Class")
    plt.ylabel("Probability")
    plt.xticks(range(n_classes))
    plt.ylim(0, 1)
    plt.show()
```
